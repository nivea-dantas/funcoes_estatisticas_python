## balanced_Random_Over_Sampler

###  Naive random over-sampling
* https://imbalanced-learn.org/stable/references/generated/imblearn.over_sampling.RandomOverSampler.html#imblearn.over_sampling.RandomOverSampler
* gera novas amostras de forma aleatória a fim de equiparar as classes do dataset.
* permite amostrar dados heterogeneos (que possuem strings)
* utiliza a  formula de smooth bootstrap para suavizar os espaços entre as amostras

### Função

```
from imblearn.over_sampling import RandomOverSampler

def balanced_Random_Over_Sampler(X,y, sampling_strategy='auto', random_state=None, shrinkage=None):
  # https://adb-6226397775262915.15.azuredatabricks.net/?o=6226397775262915#notebook/821373908545403/command/821373908545413
  '''
  X = As variáveis de x devem der as preditoras que devem estar em um formato dataframe do Pandas
  y = A variável y é a variável target a ser predita, ela precisa estar em formato de array.
  retorno: dataframe dos dados sobreamostrado e um array com os dados balanceados com o mesmo número de variáveis. 
  '''
  ros = RandomOverSampler(sampling_strategy=sampling_strategy, random_state=random_state, shrinkage=shrinkage)
  X_resampled, y_resampled = ros.fit_resample(X, y)
  print(f'Retirada de registro entre as classes usando o parametro sampling_strategy: {sampling_strategy}')
  print(f'Controlador de randomização do algoritmo: {random_state}')
  print(f'Controlador de de contração da matriz de covariância: {shrinkage}')
  print('Resampled dataset shape %s' % Counter(y_resampled))
  return X_resampled, y_resampled
```

## balanced_SMOTE

#### SMOTE
* https://imbalanced-learn.org/stable/references/generated/imblearn.over_sampling.SMOTE.html
* É uma das técnicas mais famosas quando falamos sobre over sampling.
* Equilibra a distribuição de classes aumentando aleatoriamente os exemplos das classes minoritátias, replicando-as
* Esses registros são gerados por interpolação linear
* Os registros selecionam aleatoriamente um ou mais dos K-vizinhos para cada exemplo da classe minoritária
* Diferente da ADASYN o Smote não fará distinção entre as amostras usando o a regra NN
* Conecta inliers e outliers

#### Exemplos Grágicos

```
Image(url = "https://cdn-images-1.medium.com/max/1200/1*6UFpLFl59O9e3e38ffTXJQ.png")
```

#### Função

```
from imblearn.over_sampling import SMOTE

def balanced_SMOTE(X,y, sampling_strategy='auto', random_state=None, k_neighbors=5, n_jobs=None):
  # https://adb-6226397775262915.15.azuredatabricks.net/?o=6226397775262915#notebook/821373908545403/command/97507982833368
  '''
  X = As variáveis de x devem der as preditoras que devem estar em um formato dataframe do Pandas
  y = A variável y é a variável target a ser predita, ela precisa estar em formato de array.
  retorno: dataframe dos dados sobreamostrado baseado na premissa SMOTE. Count das classes para verificação do balanceamento que foi trazido pelo algoritmo.
  '''
  sm = SMOTE(random_state=42)
  X_res, y_res = sm.fit_resample(X, y, sampling_strategy=sampling_strategy, random_state=random_state, k_neighbors=k_neighbors, n_jobs=n_jobs)
  print(f'Retirada de registro entre as classes usando o parametro sampling_strategy: {sampling_strategy}')
  print(f'Controlador de randomização do algoritmo: {random_state}')
  print(f'Número de K para construir as amostras sintéticas: {k_neighbors}')
  print(f'Numero de CPUs usados durante a cross validation: {n_jobs}')
  print('Resampled dataset shape %s' % Counter(y_res))
  return X_res, y_res
```

## balanced_SMOTENC

#### SMOTENC 
* https://imbalanced-learn.org/stable/references/generated/imblearn.over_sampling.SMOTENC.html
* Lida com variáveis categóricas
* usado para variáveis numéricas e categóricas
* é necessário fornecer essa informação no SMOTEC por meio do parâmetro categorical_features passando os indices desses recursos ou uma máscara BOOLEANA marcando esses recursos

#### Função

```
from imblearn.over_sampling import SMOTENC

def balanced_SMOTENC(X, y, categorical_features, sampling_strategy='auto', random_state=None, k_neighbors=5, n_jobs=None):
  # https://adb-6226397775262915.15.azuredatabricks.net/?o=6226397775262915#notebook/821373908545403/command/821373908545419
  '''
  X = As variáveis de x devem der as preditoras que devem estar em um formato dataframe do Pandas, pode conter variáveis categóricas do tipo objeto. 
  y = A variável y é a variável target a ser predita, ela precisa estar em formato de array com valores binarios.
  categorical_features = precisa especificar quais variáveis são categóricas. pode ser um array com os indices das variáveis target, ou uma máscara destas variáveis
  retorno: dataframe dos dados sobreamostrado com o balanceamento feito nos valores binarios da variável target. 
  '''
  sm = SMOTENC(categorical_features = categorical_features, sampling_strategy=sampling_strategy, random_state=random_state, k_neighbors=k_neighbors, n_jobs=n_jobs)
  X_res, y_res = sm.fit_resample(X, y)
  print(f'Retirada de registro entre as classes usando o parametro sampling_strategy: {sampling_strategy}')
  print(f'Controlador de randomização do algoritmo: {random_state}')
  print(f'Número de K para construir as amostras sintéticas: {k_neighbors}')
  print(f'Numero de CPUs usados durante a cross validation: {n_jobs}')
  print('Resampled dataset shape %s' % Counter(y_res))
  return X_res, y_res
```

## balanced_SMOTEN

#### SMOTEN
* https://imbalanced-learn.org/stable/references/generated/imblearn.over_sampling.SMOTEN.html
* dataset de dados categóricos
* o algoritmo pode mudar do SMOTENC em duas formas
1) o algotitmo KNN não confia na distancia euclidiana, assim, a VDM (métrica do valor da diferença) também pode ser implementado pelo paramentro ValueDifferencMetric
2) uma nova amostra é gerada onde cada valor da característica corresponde à categoria mais comum vista nas amostras vizinhas pertencentes à mesma classe. 
PS: espera-se que o modelo gere uma nova categoria junto com a classe minoritária

#### Função

```
from imblearn.over_sampling import SMOTEN

def balanced_SMOTEN(X,y, sampling_strategy='auto', random_state=None, k_neighbors=5, n_jobs=None):
  # https://adb-6226397775262915.15.azuredatabricks.net/?o=6226397775262915#notebook/821373908545403/command/97507982833451
  '''
  X = As variáveis de x devem der as preditoras que devem estar em um formato dataframe do Pandas, podem ser numéricas ou do tipo object
  y = A variável y é a variável target a ser predita, ela precisa estar em formato de array.
  retorno: dataframe dos dados sobreamostrado e um array com os dados balanceados com o mesmo número de variáveis. 
  '''
  sampler = SMOTEN(sampling_strategy=sampling_strategy, random_state=random_state, k_neighbors=k_neighbors, n_jobs=n_jobs)
  X_res, y_res = sampler.fit_resample(X, y)
  print(f'Retirada de registro entre as classes usando o parametro sampling_strategy: {sampling_strategy}')
  print(f'Controlador de randomização do algoritmo: {random_state}')
  print(f'Número de K para construir as amostras sintéticas: {k_neighbors}')
  print(f'Numero de CPUs usados durante a cross validation: {n_jobs}')
  print('Resampled dataset shape %s' % Counter(y_res))
  return X_res, y_res
```

## balanced_ADASYN

#### ADASYN
* https://imbalanced-learn.org/stable/references/generated/imblearn.over_sampling.ADASYN.html
* Baseado na premissa do SMOTE. Usa a distribuição de densidade como um critério para decidir automaticamente o numero de dados sintéticos que precisam ser gerados para cada exemplo da classe minoritária.
* irá focar em gerar  amostras próximas as amostras originais que são classificadas erroneamente usando a classificação básica do KNN
* pode se concentrar apenas em outliers

#### Função

```
from imblearn.over_sampling import ADASYN

def balanced_ADASYN(X,y, sampling_strategy='auto', random_state=None, n_neighbors=5, n_jobs=None):
  # https://adb-6226397775262915.15.azuredatabricks.net/?o=6226397775262915#notebook/821373908545403/command/821373908545425
  '''
  X = As variáveis de x devem der as preditoras que devem estar em um formato dataframe do Pandas
  y = A variável y é a variável target a ser predita, ela precisa estar em formato de array.
  retorno: dataframe dos dados sobreamostrado e um array com os dados balanceados com o mesmo número de variáveis. 
  '''
  ada = ADASYN(sampling_strategy=sampling_strategy, random_state=random_state, n_neighbors=n_neighbors, n_jobs=n_jobs)
  X_res, y_res = ada.fit_resample(X,y)
  print(f'Retirada de registro entre as classes usando o parametro sampling_strategy: {sampling_strategy}')
  print(f'Controlador de randomização do algoritmo: {random_state}')
  print(f'Número de K para construir as amostras sintéticas: {n_neighbors}')
  print(f'Numero de CPUs usados durante a cross validation: {n_jobs}')
  print('Resampled dataset shape %s' % Counter(y_res))
  return X_res, y_res
```

## balanced_borderline_SMOTE

#### BorderlineSMOTE
* https://imbalanced-learn.org/stable/references/generated/imblearn.over_sampling.BorderlineSMOTE.html
* identifica amostras proximas a superficie de decisão, e aplica o smothe somente em instancias do conjunto DANGER.
* esse conjunto será formado por instancias que possuem mais vizinhos da classe majoritária do que da classe minoritária

#### Exemplos Grágicos

```
Image(url= "https://images2.pianshen.com/262/70/701a30ae0341b39776490d65f944a12e.png")
```

#### Função

```
from imblearn.over_sampling import BorderlineSMOTE 

def balanced_borderline_SMOTE(X,y, sampling_strategy='auto', random_state=None, k_neighbors=5, n_jobs=None, m_neighbors=10, kind='borderline-1'):
  # https://adb-6226397775262915.15.azuredatabricks.net/?o=6226397775262915#notebook/821373908545403/command/821373908545429
  '''
  X = As variáveis de x devem der as preditoras que devem estar em um formato dataframe do Pandas
  y = A variável y é a variável target a ser predita, ela precisa estar em formato de array.
  retorno: dataframe dos dados sobreamostrado e um array com os dados balanceados com o mesmo número de variáveis. 
  '''
  sm = BorderlineSMOTE(sampling_strategy=sampling_strategy, random_state=random_state, k_neighbors=k_neighbors, n_jobs=n_jobs, m_neighbors=m_neighbors, kind=kind)
  X_res, y_res = sm.fit_resample(X, y)
  print(f'Retirada de registro entre as classes usando o parametro sampling_strategy: {sampling_strategy}')
  print(f'Controlador de randomização do algoritmo: {random_state}')
  print(f'Número de K para construir as amostras sintéticas: {k_neighbors}')
  print(f'Numero de CPUs usados durante a cross validation: {n_jobs}')
  print(f'Determinador da classe menor se está na condição danger: {m_neighbors}')
  print(f'Tipo do algoritmo borderline: {kind}')
  print('Resampled dataset shape %s' % Counter(y_res))
  return X_res, y_res
```

## balanced_KMeansSMOTE

#### K-Means SMOTE
* https://imbalanced-learn.org/stable/references/generated/imblearn.over_sampling.KMeansSMOTE.html
* o algoritmo utiliza 3 passos
1) clusteriza todo o input usando o algoritmo k-means
2) distribui o numero de amostras geradas por meio do cluster - selecionando clusters que tem um numero pequeno de classes e atribui mais amostras onde a distribuição é esparsa.
3) faz o oversampling com os clusters filtrados udando a Tecnica SMOTE.

#### Função

```
from imblearn.over_sampling import KMeansSMOTE

def balanced_KMeansSMOTE(X,y, sampling_strategy='auto', random_state=None, k_neighbors=2, n_jobs=None, kmeans_estimator=None, cluster_balance_threshold='auto', density_exponent='auto'):
  # https://adb-6226397775262915.15.azuredatabricks.net/?o=6226397775262915#notebook/821373908545403/command/821373908545433
  '''
  X = As variáveis de x devem der as preditoras que devem estar em um formato dataframe do Pandas
  y = A variável y é a variável target a ser predita, ela precisa estar em formato de array.
  retorno: dataframe dos dados sobreamostrado e um array com os dados balanceados com o mesmo número de variáveis. 
  '''
  sm = KMeansSMOTE(sampling_strategy=sampling_strategy, random_state=random_state, k_neighbors=k_neighbors, n_jobs=n_jobs, kmeans_estimator=kmeans_estimator, cluster_balance_threshold=cluster_balance_threshold, density_exponent=density_exponent)
  X_res, y_res = sm.fit_resample(X, y)
  print(f'Retirada de registro entre as classes usando o parametro sampling_strategy: {sampling_strategy}')
  print(f'Controlador de randomização do algoritmo: {random_state}')
  print(f'Número de K para construir as amostras sintéticas: {k_neighbors}')
  print(f'Numero de CPUs usados durante a cross validation: {n_jobs}')
  print(f'Instancia K-means ou o numero de clusters a serem usados: {kmeans_estimator}')
  print(f'Limite de clusters que serão sobreamostradas pelo SMOTE: {cluster_balance_threshold}')
  print(f'Densidade do cluster: {density_exponent}')
  print('Resampled dataset shape %s' % Counter(y_res))
  return X_res, y_res
```

## balanced_SVMSMOTE

#### SVMSMOTE
* https://imbalanced-learn.org/stable/references/generated/imblearn.over_sampling.SVMSMOTE.html
* variação do borderline SMOTE
* ao inves de usar o KNN para identificar a classificação incorreta como o borderline ela implementa o Suport Vector Machine
* A área limítrofe é aprocimada pelos vetores de suporte após o treinamento do classificador SVM no conjunto de dados original
* os dados sintéticos serão criados aleatoriamente ao longo das linhas que unem cada vetor de suporte da classe minoritária com vários de deus vizinhos mais próximos
* outra diferença do SVMSMOTE é que os dados são sintetizados longe da região de sobreposição de classes e se concentra mais em onde os dados são separados.

#### Função

```
from imblearn.over_sampling import SVMSMOTE 

def balanced_SVMSMOTE(X,y, sampling_strategy='auto', random_state=None, k_neighbors=5, n_jobs=None, m_neighbors=10, svm_estimator=None, out_step=0.5):
  # https://adb-6226397775262915.15.azuredatabricks.net/?o=6226397775262915#notebook/821373908545403/command/821373908545436
  '''
  X = As variáveis de x devem der as preditoras que devem estar em um formato dataframe do Pandas
  y = A variável y é a variável target a ser predita, ela precisa estar em formato de array.
  retorno: dataframe dos dados sobreamostrado e um array com os dados balanceados com o mesmo número de variáveis. 
  '''
  sm = SVMSMOTE(sampling_strategy=sampling_strategy, random_state=random_state, k_neighbors=k_neighbors, n_jobs=n_jobs, m_neighbors=m_neighbors, svm_estimator=svm_estimator, out_step=out_step)
  X_res, y_res = sm.fit_resample(X, y)
  print(f'Retirada de registro entre as classes usando o parametro sampling_strategy: {sampling_strategy}')
  print(f'Controlador de randomização do algoritmo: {random_state}')
  print(f'Número de K para construir as amostras sintéticas: {k_neighbors}')
  print(f'Numero de CPUs usados durante a cross validation: {n_jobs}')
  print(f'Determinador da classe menor se está na condição danger: {m_neighbors}')
  print(f'Estimador SVM: {svm_estimator}')
  print(f'Tamanho do extrapolador: {out_step}')
  print('Resampled dataset shape %s' % Counter(y_res))
  return X_res, y_res
```
