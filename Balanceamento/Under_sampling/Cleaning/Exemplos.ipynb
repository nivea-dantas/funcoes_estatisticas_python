## Carregar os dados e transformar

```
df1 = spark.read.format("csv").option("header", "true").load("dbfs:/FileStore/shared_uploads/nhmfdantas@apoioprodesp.sp.gov.br/IRIS.csv")
```

```
df1 = df1.withColumn("sepal_length", col("sepal_length").cast("double"))
df1 = df1.withColumn("sepal_Width", col("sepal_width").cast("double"))
df1 = df1.withColumn("petal_length", col("petal_length").cast("double"))
df1 = df1.withColumn("petal_width", col("petal_width").cast("double"))
```

### Desbalancear a massa de dados para utilizar as funcionalidades.

```
df1 = df1.filter(df1.sepal_length != 6.2).drop()
#.show()
```

```
df1.groupBy('species').count().show()
```

```
y = df1.select("species").withColumn("code", when(df1.species == "Iris-virginica", 0).when(df1.species == "Iris-setosa", 1).otherwise(2))
y = np.array(y.select("code").collect())
X = df1.select(["sepal_length", "sepal_Width", "petal_length", "petal_width"]).toPandas()
```

## Methods

### clean_Tomek_Links
* limpa o dataset de acordo com a detecção dos links Tomek

#### Documentação
---

https://adb-6226397775262915.15.azuredatabricks.net/?o=6226397775262915#notebook/97507982832889/command/97507982832892

#### Exemplo

```
X_blanced, y_balanced = clean_Tomek_Links(X,y)
```

### clean_Edited_Nearest_Neighbours

* Aplica um algoritmo de KNN e “edita” o conjunto de dados removendo amostras que não concordam “o suficiente” com sua vizinhança.

#### Documentação
---
https://adb-6226397775262915.15.azuredatabricks.net/?o=6226397775262915#notebook/97507982832889/command/97507982833504

#### Exemplo

```
X_blanced, y_balanced = clean_Edited_Nearest_Neighbours(X,y)
```

### edited_Repeated_Nearest_Neighbours
* Estende EditedNearestNeighbours repetindo o algoritmo várias vezes.

#### Documentação
---
https://adb-6226397775262915.15.azuredatabricks.net/?o=6226397775262915#notebook/97507982832889/command/97507982833524

#### Exemplo

```
X_blanced, y_balanced = edited_Repeated_Nearest_Neighbours(X, y)
```

### clean_All_KNN
* Faz o mesmo processo de repeated nearest neighbor, só que a cada iteração o algoritmo cria um novo NN

#### Documentação
