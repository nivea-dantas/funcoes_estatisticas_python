# Under-sampling methods

## Teoria

As funções Undersampling assumem 3 tipos de redução de dados. São eles: Remoção de registro, limpeza de registro e condensamento dos registros.
Usar esta abordagem é eficaz em situações em que a classe minoritária tem uma quantidade suficiente de exemplos, apesar do desequilíbrio severo. Por outro lado, é sempre importante considerar a possibilidade de que informações valiosas sejam excluídas à medida que as removemos aleatoriamente de nosso conjunto de dados, uma vez que não temos como detectar ou preservar os exemplos que são ricos em informações na classe da maioria.

* Existem duas formas aritméticas que regem esses processos. A geração de protótipos, para balanceamento de dados. E a Seleção de protótipos, também para baçanceamento, limpeza e condensamento.

Aqui trataremos apenas:

*  Seleção de Protótipos

Selecionarão amostras do conjunto original S. Portanto, S' é definido como |S'| < |S| e S'\subset S.

1) as técnicas de subamostragem controlada - permite uma estratégia de subamostragem na qual o número de amostras em S' é especificado pelo usuário
2) as técnicas de subamostragem de limpeza - são destinadas a limpar o espaço de características.

# Metodos

### clean_Tomek_Links

#### Tomek Links
* https://imbalanced-learn.org/stable/references/generated/imblearn.under_sampling.TomekLinks.html#imblearn.under_sampling.TomekLinks

TomekLinks detecta os chamados links do Tomek. A ligação de um Tomek entre duas amostras de diferentes classes x e y é definida de tal forma que para qualquer amostra z:

d(x, y) < d(x, z)  e  d(x, y) < d(y, z)

##### quando a distancia entre x e y for menor que a distancia de x e z e também a distancia de x e y for menor que a distancia de y e z.

onde d(.) é a distância entre as duas amostras. Em outras palavras, o link de um Tomek existe se as duas amostras são os vizinhos mais próximos uma da outra.

##### O parâmetro sampling_strategy controla qual amostra do link será removida. Por exemplo, o padrão (ou seja, sampling_strategy='auto') removerá a amostra da classe majoritária. Ambas as amostras da classe majoritária e minoritária podem ser removidas definindo sampling_strategy como 'all'.

#### Função

```
from imblearn.under_sampling import TomekLinks

def clean_Tomek_Links(X, y, sampling_strategy='auto',n_jobs= None):
  # https://adb-6226397775262915.15.azuredatabricks.net/?o=6226397775262915#notebook/97507982833199/command/97507982833237
  """
  X = As variáveis de x devem der as preditoras que devem estar em um formato dataframe do Pandas
  y = A variável y é a variável target a ser predita, ela precisa estar em formato de array.
  retorno: retorna um dataset com a exclusão dos dados em função doS links tomek, ou seja, 
           ele identifica os pontos próximos de similaridade entre as classes e assim remove aqueles cujo tem maior volume e proximidade com a outra classe.
           A função também mostra a contagem dos registros após a redução de dados. 
  """
  tl = TomekLinks(sampling_strategy=sampling_strategy, n_jobs= n_jobs)
  X_res, y_res = tl.fit_resample(X, y)
  print(f'Retirada de registro entre as classes usando o parametro sampling_strategy: {sampling_strategy}')
  print(f'Numero de CPUs usados durante a cross validation: {n_jobs}')
  print('Reorganiza o dataset %s' % Counter(y_res))
  return X_res, y_res
```

### clean_Edited_Nearest_Neighbours

#### Edited Nearest Neighbours
* https://imbalanced-learn.org/stable/references/generated/imblearn.under_sampling.EditedNearestNeighbours.html#imblearn.under_sampling.EditedNearestNeighbours

Aplica um algoritmo de KNN e “edita” o conjunto de dados removendo amostras que não concordam “o suficiente” com sua vizinhança. 
##### Para cada amostra da classe a ser subamostrada, são computados os vizinhos mais próximos e se o critério de seleção não for atendido, a amostra é removida.
##### Dois critérios de seleção estão disponíveis atualmente: 
1)  Excluirá menos variáveis (ou seja, kind_sel='mode') 
2)  Excluirá mais variáveis (ou seja, kind_sel='all') os vizinhos mais próximos devem pertencer à mesma classe que a amostra inspecionada para mantê-lo no conjunto de dados. 
Assim, isso implica que kind_sel='all' será menos conservador que kind_sel='mode', e mais amostras serão excluídas na estratégia anterior do que na última.

#### Exemplos Grágicos

```
Image(url = 'https://helloacm.com/wp-content/uploads/2016/03/2012-10-26-knn-concept.png')
```

#### Função

```
from imblearn.under_sampling import EditedNearestNeighbours

def clean_Edited_Nearest_Neighbours(X,y, kind_sel = 'all', sampling_strategy='auto', n_neighbors = 3, n_jobs = 3):
  # https://adb-6226397775262915.15.azuredatabricks.net/?o=6226397775262915#notebook/97507982833199/command/97507982833222
  
  """
  X = As variáveis de x devem der as preditoras que devem estar em um formato dataframe do Pandas
  y = A variável y é a variável target a ser predita, ela precisa estar em formato de array.
  retorno: conjunto de dados balanceado em função da técnica EditedNearestNeighbours. 
  Ele não retorna necessariamente um conjunto de dados igualmente ballanceado, mas retorna a classe e a quantidade de registros retiradas em cada uma delas.
  """
  enn = EditedNearestNeighbours(sampling_strategy=sampling_strategy, kind_sel=kind_sel, n_neighbors = n_neighbors, n_jobs = n_jobs)
  X_res, y_res = enn.fit_resample(X, y)
  print(f'Retirada de registro entre as classes usando o parametro sampling_strategy: {sampling_strategy}')
  print(f'Retirada de registro entre as classes usando o parametro kind_sel: {kind_sel}')
  print(f'O numero de vizinhos considerados para calcular a media da distancia para as amostras dos pontos minoritários: {n_neighbors}')
  print(f'Numero de CPUs usados durante a cross validation: {n_jobs}')
  print('Reorganiza o dataset %s' % Counter(y_res))
  return X_res, y_res
```

### edited_Repeated_Nearest_Neighbours

#### Repeated Edited Nearest Neighbours
* https://imbalanced-learn.org/stable/references/generated/imblearn.under_sampling.RepeatedEditedNearestNeighbours.html#imblearn.under_sampling.RepeatedEditedNearestNeighbours

Estende EditedNearestNeighbours repetindo o algoritmo várias vezes. Geralmente, repetir o algoritmo excluirá mais dados.

#### Função

```
from imblearn.under_sampling import RepeatedEditedNearestNeighbours

def edited_Repeated_Nearest_Neighbours(X, y, kind_sel = 'all', sampling_strategy='auto', n_neigbors = 3, max_iter=100, n_jobs = None):
  # https://adb-6226397775262915.15.azuredatabricks.net/?o=6226397775262915#notebook/97507982833199/command/97507982833224
  """
  X = As variáveis de x devem der as preditoras que devem estar em um formato dataframe do Pandas
  y = A variável y é a variável target a ser predita, ela precisa estar em formato de array.
  retorno: conjunto de dados balanceado em função da técnica RepeatedEditedNearestNeighbours. 
  Ele não retorna necessariamente um conjunto de dados igualmente ballanceado, mas retorna a classe e a quantidade de registros retiradas em cada uma delas.
  """
  renn = RepeatedEditedNearestNeighbours(sampling_strategy=sampling_strategy, kind_sel=kind_sel, n_neigbors = n_neigbors, max_iter=max_iter, n_jobs = n_jobs)
  X_res, y_res = renn.fit_resample(X, y)
  print(f'Retirada de registro entre as classes usando o parametro sampling_strategy: {sampling_strategy}')
  print(f'Retirada de registro entre as classes usando o parametro kind_sel: {kind_sel}')
  print(f'O numero de vizinhos considerados para calcular a media da distancia para as amostras dos pontos minoritários: {n_neighbors}')
  print(f'Número máximo de iterações que algoritmo fará {max_iter}')
  print(f'Numero de CPUs usados durante a cross validation: {n_jobs}')
  print('Reorganiza o dataset %s' % Counter(y_res))
  return X_res, y_res
```

### clean_All_KNN

#### All KNN
* https://imbalanced-learn.org/stable/references/generated/imblearn.under_sampling.AllKNN.html#imblearn.under_sampling.AllKNN

Difere do RepeatedEditedNearestNeighbours anterior desde que o numero de vizinhos do KNN cresce a cada iteração.

#### Função

```
from imblearn.under_sampling import AllKNN 

def clean_All_KNN(X,y, sampling_strategy = 'auto', n_neighbors=3, kind_sel = 'all', allow_minority = False, n_jobs = None):
  # https://adb-6226397775262915.15.azuredatabricks.net/?o=6226397775262915#notebook/97507982833199/command/97507982833226
  
  """
  X = As variáveis de x devem der as preditoras que devem estar em um formato dataframe do Pandas
  y = A variável y é a variável target a ser predita, ela precisa estar em formato de array.
  Retorno:   retorno: conjunto de dados balanceado em função da técnica KNN. 
  Ele não retorna necessariamente um conjunto de dados igualmente ballanceado, mas retorna a classe e a quantidade de registros retiradas em cada uma delas.
  """
  allknn = AllKNN(sampling_strategy = sampling_strategy, n_neighbors=n_neighbors, kind_sel = kind_sel, allow_minority = allow_minority, n_jobs = n_jobs)
  X_res, y_res = allknn.fit_resample(X, y)
  print(f'Retirada de registro entre as classes usando o parametro sampling_strategy: {sampling_strategy}')
  print(f'Retirada de registro entre as classes usando o parametro kind_sel: {kind_sel}')
  print(f'O numero de vizinhos considerados para calcular a media da distancia para as amostras dos pontos minoritários: {n_neighbors}')
  print(f'permite que as classes majoritárias se tornem classes minoritparias sem uma pausa prévia: {allow_minority}')
  print(f'Numero de CPUs usados durante a cross validation: {n_jobs}')
  print('Reorganiza o dataset %s' % Counter(y_res))
  return X_res, y_res
```

### cleaning_Neighbourhood_Rule

#### Neighbourhood Cleaning Rule
* https://imbalanced-learn.org/stable/references/generated/imblearn.under_sampling.NeighbourhoodCleaningRule.html#imblearn.under_sampling.NeighbourhoodCleaningRule

* Se concentrará na limpeza dos dados do que em condensá-los. 
* Portanto utilizará a união das amostras que serão rejeitadas entre o algoritmo edited_nearest_neighbours e o classificador 3-KNN

#### Função

```
from imblearn.under_sampling import NeighbourhoodCleaningRule

def cleaning_Neighbourhood_Rule(X,y, sampling_strategy = 'auto', n_neighbors = 3, kind_sel='all', threshold_cleaning= 0.5, n_jobs = None):
  # https://adb-6226397775262915.15.azuredatabricks.net/?o=6226397775262915#notebook/97507982833199/command/97507982833228
  # Esta função está com problemas na biblioteca interna do ibmlearn.
  """
  X = As variáveis de x devem der as preditoras que devem estar em um formato dataframe do Pandas
  y = A variável y é a variável target a ser predita, ela precisa estar em formato de array.
  Retorno: reduzirá dados limpando-os e depois fazendo a união entre as amostras encontradas no ENN e KNN. 
  """

  ncr = NeighbourhoodCleaningRule(sampling_strategy = sampling_strategy, n_neighbors = n_neighbors, kind_sel = kind_sel, threshold_cleaning= threshold_cleaning, n_jobs = n_jobs)
  X_res, y_res = ncr.fit_resample(X, y)
  print(f'Retirada de registro entre as classes usando o parametro sampling_strategy: {sampling_strategy}')
  print(f'O numero de vizinhos considerados para calcular a media da distancia para as amostras dos pontos minoritários: {n_neighbors}')
  print(f'Retirada de registro entre as classes usando o parametro kind_sel: {kind_sel}')
  print(f'Limiar usado para considerar uma classe ou não durante a limpeza após a aplicação do ENN: {threshold_cleaning}')
  print(f'Numero de CPUs usados durante a cross validation: {n_jobs}')
  print('Reorganiza o dataset %s' % Counter(y_res))
  return X_res, y_res
```
