```
from imblearn.under_sampling import TomekLinks, EditedNearestNeighbours, RepeatedEditedNearestNeighbours, AllKNN, NeighbourhoodCleaningRule
```

```
def clean_Tomek_Links(X, y, sampling_strategy='auto',n_jobs= None):
  # https://adb-6226397775262915.15.azuredatabricks.net/?o=6226397775262915#notebook/97507982833199/command/97507982833237
  """
  X = As variáveis de x devem der as preditoras que devem estar em um formato dataframe do Pandas
  y = A variável y é a variável target a ser predita, ela precisa estar em formato de array.
  retorno: retorna um dataset com a exclusão dos dados em função doS links tomek, ou seja, 
           ele identifica os pontos próximos de similaridade entre as classes e assim remove aqueles cujo tem maior volume e proximidade com a outra classe.
           A função também mostra a contagem dos registros após a redução de dados. 
  """
  tl = TomekLinks(sampling_strategy=sampling_strategy, n_jobs= n_jobs)
  X_res, y_res = tl.fit_resample(X, y)
  print(f'Retirada de registro entre as classes usando o parametro sampling_strategy: {sampling_strategy}')
  print(f'Numero de CPUs usados durante a cross validation: {n_jobs}')
  print('Balanceamento das classes no dataset %s' % Counter(y_res))
  return X_res, y_res
```

```
def clean_Edited_Nearest_Neighbours(X,y, kind_sel = 'all', sampling_strategy='auto', n_neighbors = 3, n_jobs = 3):
  # https://adb-6226397775262915.15.azuredatabricks.net/?o=6226397775262915#notebook/97507982833199/command/97507982833222
  """
  X = As variáveis de x devem der as preditoras que devem estar em um formato dataframe do Pandas
  y = A variável y é a variável target a ser predita, ela precisa estar em formato de array.
  retorno: conjunto de dados balanceado em função da técnica EditedNearestNeighbours. 
  Ele não retorna necessariamente um conjunto de dados igualmente ballanceado, mas retorna a classe e a quantidade de registros retiradas em cada uma delas.
  """
  enn = EditedNearestNeighbours(sampling_strategy=sampling_strategy, kind_sel=kind_sel, n_neighbors = n_neighbors, n_jobs = n_jobs)
  X_res, y_res = enn.fit_resample(X, y)
  print(f'Retirada de registro entre as classes usando o parametro sampling_strategy: {sampling_strategy}')
  print(f'Retirada de registro entre as classes usando o parametro kind_sel: {kind_sel}')
  print(f'O numero de vizinhos considerados para calcular a media da distancia para as amostras dos pontos minoritários: {n_neighbors}')
  print(f'Numero de CPUs usados durante a cross validation: {n_jobs}')
  print('Balanceamento das classes no dataset %s' % Counter(y_res))
  return X_res, y_res
```

```
def edited_Repeated_Nearest_Neighbours(X, y, sampling_strategy='auto', n_neighbors=3, max_iter=100, kind_sel='all', n_jobs=None): # Detalhar mais as funcionalidades
  # https://adb-6226397775262915.15.azuredatabricks.net/?o=6226397775262915#notebook/97507982833199/command/97507982833224
  """
  X = As variáveis de x devem der as preditoras que devem estar em um formato dataframe do Pandas
  y = A variável y é a variável target a ser predita, ela precisa estar em formato de array.
  retorno: conjunto de dados balanceado em função da técnica RepeatedEditedNearestNeighbours. 
  Ele não retorna necessariamente um conjunto de dados igualmente ballanceado, mas retorna a classe e a quantidade de registros retiradas em cada uma delas.
  """
  renn = RepeatedEditedNearestNeighbours(sampling_strategy=sampling_strategy, n_neighbors=n_neighbors, max_iter=max_iter, kind_sel=kind_sel, n_jobs=n_jobs)
  X_res, y_res = renn.fit_resample(X, y)
  print(f'Retirada de registro entre as classes usando o parametro sampling_strategy: {sampling_strategy}')
  print(f'Retirada de registro entre as classes usando o parametro kind_sel: {kind_sel}')
  print(f'O numero de vizinhos considerados para calcular a media da distancia para as amostras dos pontos minoritários: {n_neighbors}')
  print(f'Número máximo de iterações que algoritmo fará {max_iter}')
  print(f'Numero de CPUs usados durante a cross validation: {n_jobs}')
  print('Balanceamento das classes no dataset %s' % Counter(y_res))
  return X_res, y_res
```

```
def clean_All_KNN(X,y, sampling_strategy = 'auto', n_neighbors=3, kind_sel = 'all', allow_minority = False, n_jobs = None): # Detalhar mais as funcionalidades
  # https://adb-6226397775262915.15.azuredatabricks.net/?o=6226397775262915#notebook/97507982833199/command/97507982833226
  """
  X = As variáveis de x devem der as preditoras que devem estar em um formato dataframe do Pandas
  y = A variável y é a variável target a ser predita, ela precisa estar em formato de array.
  Retorno:   retorno: conjunto de dados balanceado em função da técnica KNN. 
  Ele não retorna necessariamente um conjunto de dados igualmente ballanceado, mas retorna a classe e a quantidade de registros retiradas em cada uma delas.
  """
  allknn = AllKNN(sampling_strategy = sampling_strategy, n_neighbors=n_neighbors, kind_sel = kind_sel, allow_minority = allow_minority, n_jobs = n_jobs)
  X_res, y_res = allknn.fit_resample(X, y)
  print(f'Retirada de registro entre as classes usando o parametro sampling_strategy: {sampling_strategy}')
  print(f'Retirada de registro entre as classes usando o parametro kind_sel: {kind_sel}')
  print(f'O numero de vizinhos considerados para calcular a media da distancia para as amostras dos pontos minoritários: {n_neighbors}')
  print(f'permite que as classes majoritárias se tornem classes minoritparias sem uma pausa prévia: {allow_minority}')
  print(f'Numero de CPUs usados durante a cross validation: {n_jobs}')
  print('Balanceamento das classes no dataset %s' % Counter(y_res))
  return X_res, y_res
```

```
def cleaning_Neighbourhood_Rule(X,y, sampling_strategy = 'auto', n_neighbors = 3, kind_sel='all', threshold_cleaning= 0.5, n_jobs = None):
  # https://adb-6226397775262915.15.azuredatabricks.net/?o=6226397775262915#notebook/97507982833199/command/97507982833228
  """
  X = As variáveis de x devem der as preditoras que devem estar em um formato dataframe do Pandas
  y = A variável y é a variável target a ser predita, ela precisa estar em formato de array.
  Retorno: reduzirá dados limpando-os e depois fazendo a união entre as amostras encontradas no ENN e KNN. 
  """

  ncr = NeighbourhoodCleaningRule(sampling_strategy=sampling_strategy, n_neighbors=n_neighbors, kind_sel=kind_sel, threshold_cleaning=threshold_cleaning, n_jobs=n_jobs)
  X_res, y_res = ncr.fit_resample(X, y)
  print(f'Retirada de registro entre as classes usando o parametro sampling_strategy: {sampling_strategy}')
  print(f'O numero de vizinhos considerados para calcular a media da distancia para as amostras dos pontos minoritários: {n_neighbors}')
  print(f'Retirada de registro entre as classes usando o parametro kind_sel: {kind_sel}')
  print(f'Limiar usado para considerar uma classe ou não durante a limpeza após a aplicação do ENN: {threshold_cleaning}')
  print(f'Numero de CPUs usados durante a cross validation: {n_jobs}')
  print('Balanceamento das classes no dataset %s' % Counter(y_res))
  return X_res, y_res
```
