# Under-sampling methods

## Teoria

As funções Undersampling assumem 3 tipos de redução de dados. São eles: Remoção de registro, limpeza de registro e condensamento dos registros.
Usar esta abordagem é eficaz em situações em que a classe minoritária tem uma quantidade suficiente de exemplos, apesar do desequilíbrio severo. Por outro lado, é sempre importante considerar a possibilidade de que informações valiosas sejam excluídas à medida que as removemos aleatoriamente de nosso conjunto de dados, uma vez que não temos como detectar ou preservar os exemplos que são ricos em informações na classe da maioria.

*  Seleção de Protótipos

Selecionarão amostras do conjunto original S. Portanto, S' é definido como |S'| < |S| e S'\subset S.

1) as técnicas de subamostragem controlada - permite uma estratégia de subamostragem na qual o número de amostras em S' é especificado pelo usuário
2) as técnicas de subamostragem de limpeza - são destinadas a limpar o espaço de características.

## condensed_undersampling

### Técnicas de subamostragem de condensamento
* As tecnicas de limpeza e condensamento dos dados não irão balancear as classes, ao invés disso, cada algoritmo implementará uma descoberta(heurística) que limpará os datasets.

# Metodos

### condensed_nearest_neighbour

#### Condensed Nearest Neighbour
* https://imbalanced-learn.org/stable/references/generated/imblearn.under_sampling.CondensedNearestNeighbour.html#imblearn.under_sampling.CondensedNearestNeighbour

* usa a regra de 1-NN dpara decidir iterativamente se um registro deve ser removido ou não.

1) pega todos os dados com menor balanceamento
2) adiciona uma amostra para o target classe (que será subamostrada) y e todas as outras amostras deste data set X
3) Pega a classe x e itera o algoritmo 1-NN
4) se a amostra nao cumprir as condições, adiciona ao grupo y, caso contrário não
5) reitera todo o conjunto x até não haver mais amostras

* é sensível a ruidos e pode adicionar dados ruidosos as amostras selecionadas

#### Função

```
from imblearn.under_sampling import CondensedNearestNeighbour 

def condensed_nearest_neighbour(X,y, sampling_strategy = 'auto', random_state = None, n_neighbors = None, n_seeds_S = 1, n_jobs= None):
  # https://adb-6226397775262915.15.azuredatabricks.net/?o=6226397775262915#notebook/821373908545887/command/821373908545944
  """
  X = As variáveis de x devem der as preditoras que devem estar em um formato dataframe do Pandas
  y = A variável y é a variável target a ser predita, ela precisa estar em formato de array.
  retorno: um dataset condensado, onde se cria um subconjunto Z do conjunto de dados X.
  """
  cnn = CondensedNearestNeighbour(sampling_strategy = sampling_strategy, random_state = random_state, n_neighbors = n_neighbors, n_seeds_S = n_seeds_S, n_jobs= n_jobs)
  X_res, y_res = cnn.fit_resample(X, y)
  print(f'Retirada de registro entre as classes usando o parametro sampling_strategy: {sampling_strategy}')
  print(f'Controlador de randomização do algoritmo: {random_state}')
  print(f'O numero de vizinhos considerados para calcular a media da distancia para as amostras dos pontos minoritários: {n_neighbors}')
  print(f'Npumero de amostras para extrair e construir o conjunto S: {n_seeds_S}')
  print(f'Numero de CPUs usados durante a cross validation: {n_jobs}')
  print('Reorganiza o dataset %s' % Counter(y_res))
  return X_res, y_res
```

### condensed_one_Side_Selection

#### OneSidedSelection
* https://imbalanced-learn.org/stable/references/generated/imblearn.under_sampling.OneSidedSelection.html#imblearn.under_sampling.OneSidedSelection

* Use TomekLinks para remover amostras barulhentas.

* depois aplicará à regra 1NN à todas as amostras e aquela que não cumprir os requisitos da regra 1NN, será adicionada a um outro dataset até nao haver mais iterações a serem realizadas

#### Função

```
from imblearn.under_sampling import OneSidedSelection
def condensed_one_Side_Selection(X,y, sampling_strategy= 'auto', random_state = None, n_neighbors = None, n_seeds_S = 1, n_jobs= None):
  # https://adb-6226397775262915.15.azuredatabricks.net/?o=6226397775262915#notebook/821373908545887/command/821373908545948
  
  """
  X = As variáveis de x devem der as preditoras que devem estar em um formato dataframe do Pandas
  y = A variável y é a variável target a ser predita, ela precisa estar em formato de array.
  Retorno: Um conjunto de dados condensado de acordo com as regras estabelecidas pelo algoritmo one side selection.
  """

  oss = OneSidedSelection(sampling_strategy= sampling_strategy, random_state = random_state, n_neighbors = n_neighbors, n_seeds_S = n_seeds_S, n_jobs= n_jobs)
  X_res, y_res = oss.fit_resample(X, y)
  print(f'Retirada de registro entre as classes usando o parametro sampling_strategy: {sampling_strategy}')
  print(f'Controlador de randomização do algoritmo: {random_state}')
  print(f'O numero de vizinhos considerados para calcular a media da distancia para as amostras dos pontos minoritários: {n_neighbors}')
  print(f'Npumero de amostras para extrair e construir o conjunto S: {n_seeds_S}')
  print(f'Numero de CPUs usados durante a cross validation: {n_jobs}')
  print('Reorganiza o dataset %s' % Counter(y_res))
  return X_res, y_res
```
