```
from imblearn.under_sampling import CondensedNearestNeighbour, OneSidedSelection
```

```
def condensed_nearest_neighbour(X,y, sampling_strategy = 'auto', random_state = None, n_neighbors = None, n_seeds_S = 1, n_jobs= None):
  # https://adb-6226397775262915.15.azuredatabricks.net/?o=6226397775262915#notebook/821373908545887/command/821373908545944
  """
  X = As variáveis de x devem der as preditoras que devem estar em um formato dataframe do Pandas
  y = A variável y é a variável target a ser predita, ela precisa estar em formato de array.
  retorno: um dataset condensado, onde se cria um subconjunto Z do conjunto de dados X.
  """
  cnn = CondensedNearestNeighbour(sampling_strategy = sampling_strategy, random_state = random_state, n_neighbors = n_neighbors, n_seeds_S = n_seeds_S, n_jobs= n_jobs)
  X_res, y_res = cnn.fit_resample(X, y)
  print(f'Retirada de registro entre as classes usando o parametro sampling_strategy: {sampling_strategy}')
  print(f'Controlador de randomização do algoritmo: {random_state}')
  print(f'O numero de vizinhos considerados para calcular a media da distancia para as amostras dos pontos minoritários: {n_neighbors}')
  print(f'Npumero de amostras para extrair e construir o conjunto S: {n_seeds_S}')
  print(f'Numero de CPUs usados durante a cross validation: {n_jobs}')
  print('Reorganiza o dataset %s' % Counter(y_res))
  return X_res, y_res
```

```
def condensed_one_Side_Selection(X,y, sampling_strategy= 'auto', random_state = None, n_neighbors = None, n_seeds_S = 1, n_jobs= None):
  # https://adb-6226397775262915.15.azuredatabricks.net/?o=6226397775262915#notebook/821373908545887/command/821373908545948
  """
  X = As variáveis de x devem der as preditoras que devem estar em um formato dataframe do Pandas
  y = A variável y é a variável target a ser predita, ela precisa estar em formato de array.
  Retorno: Um conjunto de dados condensado de acordo com as regras estabelecidas pelo algoritmo one side selection.
  """

  oss = OneSidedSelection(sampling_strategy= sampling_strategy, random_state = random_state, n_neighbors = n_neighbors, n_seeds_S = n_seeds_S, n_jobs= n_jobs)
  X_res, y_res = oss.fit_resample(X, y)
  print(f'Retirada de registro entre as classes usando o parametro sampling_strategy: {sampling_strategy}')
  print(f'Controlador de randomização do algoritmo: {random_state}')
  print(f'O numero de vizinhos considerados para calcular a media da distancia para as amostras dos pontos minoritários: {n_neighbors}')
  print(f'Npumero de amostras para extrair e construir o conjunto S: {n_seeds_S}')
  print(f'Numero de CPUs usados durante a cross validation: {n_jobs}')
  print('Reorganiza o dataset %s' % Counter(y_res))
  return X_res, y_res
```
