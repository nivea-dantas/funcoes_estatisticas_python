```
from imblearn.under_sampling import RandomUnderSampler, NearMiss, ClusterCentroids, InstanceHardnessThreshold
```

```
def balanced_Cluster_Centroids(X, y, sampling_strategy = 'auto', random_state = None, estimator = None, voting = 'auto', n_jobs = None):
  # https://adb-6226397775262915.15.azuredatabricks.net/?o=6226397775262915#notebook/97507982833239/command/1121831065218961
  """
  X = As variáveis de x devem der as preditoras que devem estar em um formato dataframe do Pandas
  y = A variável y é a variável target a ser predita, ela precisa estar em formato de array.
  retorno: dataframe dos dados e um array com os dados balanceados com o mesmo número de variáveis. 
  """
  cc = ClusterCentroids(sampling_strategy = sampling_strategy, random_state = random_state, estimator = estimator, voting = voting, n_jobs = n_jobs)
  X_res, y_res = cc.fit_resample(X, y)
  print(f'Retirada de registro entre as classes usando o parametro sampling_strategy: {sampling_strategy}')
  print(f'Controlador de randomização do algoritmo: {random_state}')
  print(f'Escolha de estimador, por padrão é o K-means: {estimator}')
  print(f'Estratégia para gerar novas amostras: {voting}')
  print(f'Numero de CPUs usados durante a cross validation: {n_jobs}')
  print('Balanceamento das classes no dataset %s' % Counter(y_res))
  return X_res, y_res
```

```
def balanced_Random_Under_Sampler(X,y, sampling_strategy = 'auto', random_state = None, replacement = False):
  # https://adb-6226397775262915.15.azuredatabricks.net/?o=6226397775262915#notebook/97507982833239/command/1121831065219029
  """
  X = As variáveis de x devem der as preditoras que devem estar em um formato dataframe do Pandas
  y = A variável y é a variável target a ser predita, ela precisa estar em formato de array.
  retorno: dataset balanceado com o mesmo numero de registros para cada classe e um array contendo a classificação dos dados.
  """
  rus = RandomUnderSampler(sampling_strategy = sampling_strategy, random_state = random_state, replacement = replacement)
  X_res, y_res = rus.fit_resample(X, y)
  print(f'Retirada de registro entre as classes usando o parametro sampling_strategy: {sampling_strategy}')
  print(f'Controlador de randomização do algoritmo: {random_state}')
  print(f'Informa se a amostra possui ou não substituição: {replacement}')
  print('Balanceamento das classes no dataset %s' % Counter(y_res))
  return X_res, y_res
```

```
def balanced_NearMiss(X, y,sampling_strategy = 'auto', version = 1, n_neighbors = 3, n_neighbors_ver3 = 3, n_jobs= None):
  # https://adb-6226397775262915.15.azuredatabricks.net/?o=6226397775262915#notebook/97507982833239/command/1121831065219033
  
  """
  X = As variáveis de x devem der as preditoras que devem estar em um formato dataframe do Pandas
  y = A variável y é a variável target a ser predita, ela precisa estar em formato de array.
  retorno: redução de dados em função do balanceamento ao numero menor de classe encontrarda no seu dataset
  """
  nm = NearMiss(sampling_strategy = sampling_strategy, version = version, n_neighbors = n_neighbors, n_neighbors_ver3 = n_neighbors_ver3, n_jobs= n_jobs)
  X_res, y_res = nm.fit_resample(X, y)
  print(f'Retirada de registro entre as classes usando o parametro sampling_strategy: {sampling_strategy}')
  print(f'Versão do algoritmo near miss: {version}')
  print(f'O numero de vizinhos considerados para calcular a media da distancia para as amostras dos pontos minoritários: {n_neighbors}')
  print(f'Número de vizinhos selecionados para a criação de subsets: {n_neighbors_ver3}')
  print(f'Numero de CPUs usados durante a cross validation: {n_jobs}')
  print('Balanceamento das classes no dataset %s' % Counter(y_res))
  return X_res, y_res
```

```
def balanced_Instance_Hardness_Threshold(X,y, estimator=None, sampling_strategy='auto', random_state=None, cv=5, n_jobs=None):
  # https://adb-6226397775262915.15.azuredatabricks.net/?o=6226397775262915#notebook/97507982833239/command/1121831065218798
  """
  X = As variáveis de x devem der as preditoras que devem estar em um formato dataframe do Pandas
  y = A variável y é a variável target a ser predita, ela precisa estar em formato de array.
  Retorno:  
  """
  iht = InstanceHardnessThreshold(estimator=estimator, sampling_strategy=sampling_strategy, random_state=random_state, cv=cv, n_jobs=n_jobs)
  X_res, y_res = iht.fit_resample(X, y)
  print(f'Escolha de estimador, por padrão é o K-means: {estimator}')
  print(f'Retirada de registro entre as classes usando o parametro sampling_strategy: {sampling_strategy}')
  print(f'Controlador de randomização do algoritmo: {random_state}')
  print(f' Numero de dobras a serem usadas ao estimar a dureza da instancia: {cv}')
  print(f'Numero de CPUs usados durante a cross validation: {n_jobs}')
  print('Balanceamento das classes no dataset %s' % Counter(y_res))
  return X_res, y_res
```
