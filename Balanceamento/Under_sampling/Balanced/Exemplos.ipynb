## Carregar os dados e transformar

```
df1 = spark.read.format("csv").option("header", "true").load("dbfs:/FileStore/shared_uploads/nhmfdantas@apoioprodesp.sp.gov.br/IRIS.csv")
```

```
df1 = df1.withColumn("sepal_length", col("sepal_length").cast("double"))
df1 = df1.withColumn("sepal_Width", col("sepal_width").cast("double"))
df1 = df1.withColumn("petal_length", col("petal_length").cast("double"))
df1 = df1.withColumn("petal_width", col("petal_width").cast("double"))
```

### Desbalancear a massa de dados para utilizar as funcionalidades.

```
df1 = df1.filter(df1.sepal_length != 6.2).drop()
#.show()
```

```
df1.groupBy('species').count().show()
```

```
y = df1.select("species").withColumn("code", when(df1.species == "Iris-virginica", 0).when(df1.species == "Iris-setosa", 1).otherwise(2))
y = np.array(y.select("code").collect())
X = df1.select(["sepal_length", "sepal_Width", "petal_length", "petal_width"]).toPandas()
```

## Methods

### balanced_Cluster_Centroids
* Faz uso de K-means para reduzir o número de amostras.

#### Documentação
---

#### Exemplo

```
X_balanced, y_balanced = balanced_Cluster_Centroids(X, y)
```

### balanced_Random_Under_Sampler
* Balanceia o data set selecionando os dados aleatoriamente de uma maneira fácil e rapida.

#### Documentação
---
https://adb-6226397775262915.15.azuredatabricks.net/?o=6226397775262915#notebook/97507982832876/command/1121831065218966

#### Exemplo

```
X_balanced, y_balanced = balanced_Random_Under_Sampler(X,y)
```

### balanced_NearMiss
* Este algoritmo verifica a distribuição das classes usando o algoritmo nearest neigthbor, e randomicamente elimina dados que estão em maior numero dentro de uma ou mais classes.

* Pode ser implementado de 3 formas selecionadas pelo parametro "version".

#### Documentação
---
https://adb-6226397775262915.15.azuredatabricks.net/?o=6226397775262915#notebook/97507982832876/command/1121831065218962

#### Exemplo

```
X_balanced, y_balanced = balanced_NearMiss(X, y, version = 1)
```

```
X_balanced, y_balanced = balanced_NearMiss(X, y, version = 2)
```

```
X_balanced, y_balanced = balanced_NearMiss(X, y, version = 3)
```
