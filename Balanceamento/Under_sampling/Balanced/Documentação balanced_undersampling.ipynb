# Under-sampling methods

## Teoria

As funções Undersampling assumem 3 tipos de redução de dados. São eles: Remoção de registro, limpeza de registro e condensamento dos registros.
Usar esta abordagem é eficaz em situações em que a classe minoritária tem uma quantidade suficiente de exemplos, apesar do desequilíbrio severo. Por outro lado, é sempre importante considerar a possibilidade de que informações valiosas sejam excluídas à medida que as removemos aleatoriamente de nosso conjunto de dados, uma vez que não temos como detectar ou preservar os exemplos que são ricos em informações na classe da maioria.

* Existem duas formas aritméticas que regem esses processos. A geração de protótipos, para balanceamento de dados. E a Seleção de protótipos, também para baçanceamento, limpeza e condensamento.

* Geração de Protótipos - Engloba apenas a técnica de Custer Centróides

Dado um conjunto de dados original S, algoritmos de geração de protótipos irão gerar um novo conjunto S' onde |S'| < |S| e S' \não\subconjunto
S. Em outras palavras, a técnica de geração de protótipos reduzirá o número de amostras nas classes alvo, mas as amostras restantes são geradas — e não selecionadas — a partir do conjunto original.

*  Seleção de Protótipos

Selecionarão amostras do conjunto original S. Portanto, S' é definido como |S'| < |S| e S'\subset S.

1) as técnicas de subamostragem controlada - permite uma estratégia de subamostragem na qual o número de amostras em S' é especificado pelo usuário
2) as técnicas de subamostragem de limpeza - são destinadas a limpar o espaço de características.
2) as técnicas de subamostragem de Condensamento - resume os datasets à uma classe ou não, depende da setagem gos hiperparâmetros.

##### balanced_Cluster_Centroids

###### Cluster Centroids
Faz uso de K-means para reduzir o número de amostras. Portanto, cada classe será sintetizada com os centroides do método K-means em vez das amostras originais.
PS = reduz o numero de amostras de acordo com os centróides gerados pelo algoritmo k-means.

###### Função

```
from imblearn.under_sampling import ClusterCentroids 

def balanced_Cluster_Centroids(X, y, sampling_strategy = 'auto', random_state = None, estimator = None, voting = 'auto', n_jobs = None):
  # https://adb-6226397775262915.15.azuredatabricks.net/?o=6226397775262915#notebook/97507982833239/command/1121831065218961
  """
  X = As variáveis de x devem der as preditoras que devem estar em um formato dataframe do Pandas
  y = A variável y é a variável target a ser predita, ela precisa estar em formato de array.
  retorno: dataframe dos dados e um array com os dados balanceados com o mesmo número de variáveis. 
  """
  cc = ClusterCentroids()
  X_res, y_res = cc.fit_resample(X, y, sampling_strategy = sampling_strategy, random_state = random_state, estimator = estimator, voting = n_jobs, n_jobs = n_jobs)
  print(f'Retirada de registro entre as classes usando o parametro sampling_strategy: {sampling_strategy}')
  print(f'Controlador de randomização do algoritmo: {random_state}')
  print(f'Escolha de estimador, por padrão é o K-means: {estimator}')
  print(f'Estratégia para gerar novas amostras: {voting}')
  print(f'Numero de CPUs usados durante a cross validation: {n_jobs}')
  print('Reorganiza o dataset %s' % Counter(y_res))
  return X_res, y_res
```

##### balanced_Random_Under_Sampler

###### Random Under Sampler
* https://imbalanced-learn.org/stable/references/generated/imblearn.under_sampling.RandomUnderSampler.html#imblearn.under_sampling.RandomUnderSampler

* Balanceia o data set selecionando os dados aleatoriamente de uma maneira fácil e rapida.
* Também consegue utilizar dados heterogeneos (strings)

###### Função

```
from imblearn.under_sampling import RandomUnderSampler

def balanced_Random_Under_Sampler(X,y, sampling_strategy = 'auto', random_state = None, replacement = False):
  # https://adb-6226397775262915.15.azuredatabricks.net/?o=6226397775262915#notebook/97507982833239/command/1121831065219029
  """
  X = As variáveis de x devem der as preditoras que devem estar em um formato dataframe do Pandas
  y = A variável y é a variável target a ser predita, ela precisa estar em formato de array.
  retorno: dataset balanceado com o mesmo numero de registros para cada classe e um array contendo a classificação dos dados.
  """
  rus = RandomUnderSampler(sampling_strategy = sampling_strategy, random_state = random_state, replacement = replacement)
  X_res, y_res = rus.fit_resample(X, y)
  print(f'Retirada de registro entre as classes usando o parametro sampling_strategy: {sampling_strategy}')
  print(f'Controlador de randomização do algoritmo: {random_state}')
  print(f'Informa se a amostra possui ou não substituição: {replacement}')
  print('Reorganiza o dataset %s' % Counter(y_res))
  return X_res, y_res
```

##### balanced_NearMiss

###### NearMiss
* https://imbalanced-learn.org/stable/references/generated/imblearn.under_sampling.NearMiss.html#imblearn.under_sampling.NearMiss

Este algoritmo verifica a distribuição das classes usando o algoritmo nearest neigthbor, e randomicamente elimina dados que estão em maior numero dentro de uma ou mais classes.

Pode ser implementado de 3 formas selecionadas pelo parametro "version".

* o Calculo para clusterização será a distancia Euclidiana.

##### Near miss 1 - seleciona as amostras com maior balanceamento, e encontra as variáveis que possuiem MENOR distancia media (do ponto central KNN) da classe alvo de balanceamento. Pode ser alterado pela presença de ruido - implica na seleção da classe alvo em torno dessas amostras.
##### Near Miss 2 - seleciona as amostras com maior balanceamento, e encontra as variáveis que possuiem MAIOR distancia media (do ponto central KNN) da classe alvo de balanceamento. Pode sofrer alteração de ruido se o dataset obtiver presença de outliers.
##### Near Miss 3 - Possui duas etapas: 1) Para cada amostra com menor numero de registros, seus vizinhos mais próximos serão mantidos. 2) amostras que serão selecionadas para a subamostragem serão aquelas com maior distancia do ponto central do KNN. É o processo que é menos afetada pelo ruido por causa do primeiro processo.

###### Função

```
from imblearn.under_sampling import NearMiss

def balanced_NearMiss(X, y,sampling_strategy = 'auto', version = 1, n_neighbors = 3, n_neighbors_ver3 = 3, n_jobs= None):
  # https://adb-6226397775262915.15.azuredatabricks.net/?o=6226397775262915#notebook/97507982833239/command/97507982833255
  
  """
  X = As variáveis de x devem der as preditoras que devem estar em um formato dataframe do Pandas
  y = A variável y é a variável target a ser predita, ela precisa estar em formato de array.
  retorno: redução de dados em função do balanceamento ao numero menor de classe encontrarda no seu dataset
  """
  nm = NearMiss()
  X_res, y_res = nm.fit_resample(X, y, sampling_strategy = sampling_strategy, version = version, n_neighbors = n_neighbors, n_neighbors_ver3 = n_neighbors_ver3, n_jobs= n_jobs)
  print(f'Retirada de registro entre as classes usando o parametro sampling_strategy: {sampling_strategy}')
  print(f'Versão do algoritmo near miss: {version}')
  print(f'O numero de vizinhos considerados para calcular a media da distancia para as amostras dos pontos minoritários: {n_neighbors}')
  print(f'Número de vizinhos selecionados para a criação de subsets: {n_neighbors_ver3}')
  print(f'Numero de CPUs usados durante a cross validation: {n_jobs}')
  print('Reorganiza o dataset %s' % Counter(y_res))
  return X_res, y_res
```

##### balanced_Instance_Hardness_Threshold

###### Instance Hardness Threshold
* https://imbalanced-learn.org/stable/references/generated/imblearn.under_sampling.InstanceHardnessThreshold.html#imblearn.under_sampling.InstanceHardnessThreshold

É um algoritmo específico onde um classificador é treinado nos dados e as amostras com probabilidades mais baixas são removidas.
Esta classe tem 2 parâmetros importantes. 
* O estimador aceitará qualquer classificador scikit-learn que tenha um método predict_proba. 
* O treinamento do classificador é realizado usando uma validação cruzada e o parâmetro cv pode definir o número de dobras a serem usadas.

###### Função

```
from imblearn.under_sampling import InstanceHardnessThreshold


def balanced_Instance_Hardness_Threshold(X,y, estimator= None, sampling_strategy= 'auto', random_state = None, cv=5, n_jobs = None):
  # https://adb-6226397775262915.15.azuredatabricks.net/?o=6226397775262915#notebook/97507982833239/command/1121831065218798
  """
  X = As variáveis de x devem der as preditoras que devem estar em um formato dataframe do Pandas
  y = A variável y é a variável target a ser predita, ela precisa estar em formato de array.
  estimator = é necessário colocar tipo de modelo que irá fazer a separação.
  Retorno:  
  """
  iht = InstanceHardnessThreshold(estimator= estimator, sampling_strategy= sampling_strategy, random_state = random_state, cv=cv, n_jobs = n_jobs)
  X_res, y_res = iht.fit_resample(X, y)
  print(f'Escolha de estimador, por padrão é o K-means: {estimator}')
  print(f'Retirada de registro entre as classes usando o parametro sampling_strategy: {sampling_strategy}')
  print(f'Controlador de randomização do algoritmo: {random_state}')
  print(f' Numero de dobras a serem usadas ao estimar a dureza da instancia: {cv}')
  print(f'Numero de CPUs usados durante a cross validation: {n_jobs}')
  print('Reorganiza o dataset %s' % Counter(y_res))
  return X_res, y_res
```
