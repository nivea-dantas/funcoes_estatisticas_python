```python
#remoção usando IQR
def remove_outlier_IQR(df):
    colunas = df.columns
    df = df.select_dtypes(exclude= 'object')
        
    for i in df:
        df_out_remove = df[i]
        Q1 = np.percentile(df_out_remove, 25)
        Q3 = np.percentile(df_out_remove, 75)
        C   = 1.5
        IIQ = Q3 - Q1
        LI  = Q1 - C*IIQ
        LS  = Q3 + C*IIQ
        df_teste3 = df[df['PM10'] <= LS]
        df_teste3 = df_teste3[df['PM10'] >= LI]
    df_teste3 = pd.DataFrame(df_teste3)
    
    return df_teste3
```
