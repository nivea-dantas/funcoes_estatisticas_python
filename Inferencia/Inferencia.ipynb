```python
def inferencia_proporção(casos_sucesso,df, z_teste):
    casos_sucesso = casos_sucesso
    tamanho_da_amostra = len(df.index)
    proporcao = casos_sucesso / tamanho_da_amostra
    #a = 
    # Testa pressupostos
    if (tamanho_da_amostra * proporcao > 10) & (tamanho_da_amostra * (1 - proporcao) > 10):
        print('Pressupostos checados, proporções validadas!')
        erro_padrao = proporcao * (1 - proporcao)/tamanho_da_amostra
        erro_padrao = sqrt(erro_padrao)
        estimativa_superior = float((proporcao + z_teste * erro_padrao)* 100)
        estimativa_inferior = float((proporcao - z_teste * erro_padrao)* 100)
    else:
        print('A amostra não está condizente com os pressupostos')      
    return (f'O inervalo de confiança é de [{estimativa_inferior}% ; {estimativa_superior}%]')
```

```python
#calcula intervalo - média da população com disvio padrão conhecido
# podemos calcular se o dP é conhecido ou não
#possiveis respostas = a média do caso acontecer é entre x e y, com z% de confiança.
```

```python
def inferencia_media_std_populacional(desvio_padrão_populacao, df, z_teste):
    #pressupostos
    inferencias = []
    tamanho_da_amostra = len(df.index)
    
    if tamanho_da_amostra >= 30:
        print('podemos seguir com a análise')
    for colunas in df:
        df_teste = df[colunas]
        media = df_teste.mean()
        erro_padrao = desvio_padrão_populacao/sqrt(tamanho_da_amostra)
        estimativa_superior = media + z_teste * erro_padrao
        estimativa_inferior = media - z_teste * erro_padrao
        inferencia = f'O inervalo de confiança para a coluna {colunas} é de [{estimativa_inferior} ; {estimativa_superior}]'
        inferencias.append(inferencia)
    return inferencias
```

```python
def inferencial_amostral(df,T_teste): 
    inferencias = [] 
    tamanho_da_amostra = len(df.index) 
    for colunas in df: 
        df_teste = df[colunas]
        media = df_teste.mean() 
        desvio_padrao_amostral = df_teste.std()
        margem_erro = desvio_padrao_amostral/sqrt(tamanho_da_amostra) 
        estimativa_superior = media + T_teste * margem_erro 
        estimativa_inferior = media - T_teste * margem_erro 
        inferencia = f'O inervalo para a coluna {colunas} de confiança é de [ {estimativa_inferior} ; {estimativa_superior} ]' 
        inferencias.append(inferencia) 
    return inferencias
```
