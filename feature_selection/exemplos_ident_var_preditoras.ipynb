##### Análise pós Modelo

```
#unindo valiáveis preditoras com a variável target
target = y.to_frame(name='Target')
target['key'] = [i + 1 for i in  range(0, len(target))]
preditoras = preditoras.toPandas()
categoricas = preditoras.select_dtypes('object')
numericas = preditoras.drop(categoricas, axis = 1)
numericas['key'] = [i + 1 for i in range(0, len(numericas))]
gogo = numericas.set_index('key').join(target.set_index('key'), on = 'key')

# verificando a multicolinearidade
vif_data1 = pd.DataFrame()
vif_data1["feature"] = gogo.columns
vif_data1["VIF"] = [variance_inflation_factor(gogo.values, i)
                          for i in range(len(gogo.columns))]
comp = pd.merge(vif_data1, vif_data, how = 'outer', on = 'feature')
comp = pd.merge(comp, juntar_compara, how = 'outer', on = 'feature')

# criando correlação do modelo
corr_modelo = gogo.corr()
corr_modelo = corr_modelo.reset_index(level = 0)
corr_modelo = corr_modelo.rename(columns = {'index': 'feature'})
corr_modelo = corr_modelo[['feature', 'Target']]

#merge correlação do modelo à tabela de comparação
comp = pd.merge(comp, corr_modelo, how = 'inner', on = 'feature')
comp = comp.rename(columns = {'VIF_x': 'VIF_Modelo', 'VIF_y': 'VIF_dataframe', 'TARGET': 'Correlação_dataframe', 'Target': 'Correlação_Modelo'})
display(comp)
```
